## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
#f0 moet kleiner zijn dan 2.5 Hz
C = 246
m = 1
kf = np.sqrt(C/m)/(2*np.pi)
print(kf)
```

```python
f = np.linspace(0,20,100)
Ver = np.abs(1/(1-f**2/kf**2))
plt.plot(f,Ver, label = 'Verhouding')
plt.xlabel('Frequentie ($Hz$)')
plt.ylabel('Verhouding')
plt.xscale('log')
plt.yscale('log')
plt.legend()
plt.show()
```
