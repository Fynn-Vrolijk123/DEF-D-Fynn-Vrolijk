```python
# Schrijf hier een python functie die de capaciteit van een plaatcondensator uitrekent. Zorg dat de functie
# als input de ontwerpparameters van de vlakke plaat formule accepteert (oppervlakte, afstand en diëlektrische
# constante) en als output de capaciteit van de vlakke plaat condensator met die ontwerpparameters.
# Je gaat deze formule de rest van de dag gebruiken: controleer hem goed!
import numpy as np
f = 1e5
Test = 1e-5/(f*(16*2*np.pi*np.cos(2*np.pi*f)+1))
print(Test)
def vlakkeplaat(epsilon,A,d):
    C = epsilon*A/d
    return C

# Optioneel: schrijf1  je functie zodanig dat je arrays als input kan geven en (dus) ook arrays als output.
```

𝐶=𝜀*𝐴/𝑑
Omdat in ons experiment de afstand 
d constant blijft en alleen het overlapoppervlak verandert, geldt:
A=bx
Dus C∝x waar x de overlaplengte is.

Uit de metingen blijkt dat de capaciteit kwadratisch afneemt met de aangelegde kracht. Dit betekent dat mechanisch geldt:
F∝x	​
Dit kan worden verklaard doordat de tegenwerkende kracht (wrijving/adhesie) afneemt naarmate het contactoppervlak kleiner wordt, waardoor de benodigde trekkracht niet lineair maar niet-lineair met de overlap samenhangt.
Combineren van beide relaties geeft:

C(F)=C_0−a*F^2

De gebruikte fit is dus gebaseerd op:
De formule voor de capaciteit van een plaatcondensator.
Een niet-lineair mechanisch model waarbij de kracht evenredig is met de wortel van de overlap.
